## Evaluate on the `Test` dataset

```python
import pandas as pd
import json
test_dataset_path = "/content/mediasearch/MSR-VTT/msrvtt_test_1k.json"

test_dataset = pd.read_json(test_dataset_path)
test_dataset.head()
```

```python
test_dataset["video_path"] = test_dataset["video_id"].apply(lambda x: f"/content/mediasearch/MSR-VTT/MSRVTT_Videos/video/{x}.mp4")
test_dataset.head()
```

```python

from mediasearch.vit import VideoQuery
config = {
        "model_name": "ViT-B/32",
        "frame_rate": 3,
        "threshold": 0.25,
        "debug": False
}

video_search = VideoQuery(**config)
print(video_search.cash)
```

```python
%%time
video_search.insert_videos(videos_path=list(test_dataset["video_path"]))
```

```python
video_search.device
```

```python

from mediasearch.vit import VideoQuery
config = {
        "model_name": "ViT-B/32",
        "frame_rate": 3,
        "threshold": 0.25,
        "debug": False,
        "cash":"/content/drive/MyDrive/mediasearch/embeddings.h5"
}
video_search = VideoQuery(**config)
```

```python
test_dataset.shape
```

```python

from tqdm import tqdm
score = 0
for row in tqdm(test_dataset.iterrows()):
  query = row[1]["caption"]
  video_path = row[1]["video_path"]
  results = video_search.search(query) # then you can query
  if video_path in results.keys():
    score +=1
print(f"Out if the {test_dataset.shape[0]} videos, {score} videos matched correctly")
```

```python
import h5py

embeddings_file = "/content/drive/MyDrive/mediasearch/embeddings.h5"
assets_path = "/content/mediasearch/"

with h5py.File(embeddings_file, "r+") as f:

    for key in f.keys():
        # Read the current video path
        video = f[key]["video"][0].decode("utf-8")

        # Replace relative path with absolute path
        new_video = video.replace("/content/mediasearch/", "/content/")

        # Update the dataset
        f[key]["video"][0] = new_video.encode("utf-8")

        print(f"  {video} -> {new_video}")
```
